```python
import polars as pl
from src.experiments.utils import make_experiments_binary_perceptrone, polars_dict_mapper
from src.data.preprocessing.preprocessor import RuleTransform
```

```python
def prepare_heart_dataset(df: pl.DataFrame) -> pl.DataFrame:
    map_dict = {
        "AgeCategory": {
            "18-24": 0,
            "25-29": 1,
            "30-34": 2,
            "35-39": 3,
            "40-44": 4,
            "45-49": 5,
            "50-54": 6,
            "55-59": 7,
            "60-64": 8,
            "65-69": 9,
            "70-74": 10,
            "75-79": 11,
            "80 or older": 12,
        },
        "GenHealth": {
            "Poor": 0,
            "Fair": 1,
            "Good": 2,
            "Very good": 3,
            "Excellent": 4,
        },
    }
    df = polars_dict_mapper(df=df, map_dict=map_dict)
    return df

df = pl.read_csv("../data/heart_2020_cleaned.csv")
df = prepare_heart_dataset(df=df)
rules_tranform = [
    RuleTransform(
        transformer="BinaryTarget",
        columns=[
            "HeartDisease",
        ],
    ),
    RuleTransform(
        transformer="PowerTransformerFeature",
        columns=[
            "BMI",
            "PhysicalHealth",
            "MentalHealth",
            "AgeCategory",
            "GenHealth",
            "SleepTime",
        ],
    ),
    RuleTransform(
        transformer="BinaryFeature",
        columns=[
            "Smoking",
            "AlcoholDrinking",
            "Stroke",
            "DiffWalking",
            "Sex",
            "Diabetic",
            "PhysicalActivity",
            "Asthma",
            "KidneyDisease",
            "SkinCancer",
        ],
    ),
    RuleTransform(
        transformer="OneHotEncoder",
        columns=[
            "Race",
        ],
    ),
]
results = make_experiments_binary_perceptrone(
    path_to_save="../experiments/heart_dataset/",
    df=df,
    rules_tranform=rules_tranform,
    n_hidden_layers=[1, 2],
    dims_hidden=[1, 2, 4, 8, 16, 32, 64, 128, 256, 512, 1024, 2048],
    random_seeds=list(range(30)),
    f_act="ELU",
    batch_size_inference_bayesian=8,
    sample_size_inference_bayesian=1024,
    
)
```
